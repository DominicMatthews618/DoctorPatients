```python
import csv
import random
from faker import Faker
from datetime import datetime, timedelta

fake = Faker()

# Constants
genders = ['Male', 'Female']
primary_physicians = [f"Dr. {name}" for name in ['Smith', 'Johnson', 'Williams', 'Brown', 'Jones', 'Miller', 'Davis', 'Garcia']]
visit_reasons = ['Checkup', 'Flu symptoms', 'Follow-up', 'Skin rash', 'Allergy', 'Back pain']
diagnoses = ['Hypertension', 'Diabetes', 'Flu', 'Asthma', 'Healthy', 'Eczema']
medications = ['Lisinopril', 'Metformin', 'Ibuprofen', 'Albuterol', 'None']
allergies = ['Penicillin', 'Peanuts', 'Latex', 'Pollen', 'None']
chronic_conditions = ['Hypertension', 'Diabetes', 'Asthma', 'None']
insurance_providers = ['Aetna', 'Blue Cross', 'UnitedHealth', 'Kaiser Permanente', 'None']
billing_statuses = ['Paid', 'Pending', 'Overdue']
payment_methods = ['Credit Card', 'Cash', 'Insurance', 'Online Transfer']
risk_levels = ['Low', 'Medium', 'High']
treatment_outcomes = ['Resolved', 'Ongoing', 'Referred']

def generate_patient_data_with_weights(num_records=575):
    records = []

    for i in range(1, num_records + 1):
        dob = fake.date_of_birth(minimum_age=18, maximum_age=99)
        visit_date = fake.date_between(start_date='-1y', end_date='today')
        last_visit = fake.date_between(start_date='-2y', end_date=visit_date)
        next_appointment = visit_date + timedelta(days=random.randint(7, 90)) if random.random() < 0.5 else ''

        # Weighted diagnosis: 40% Healthy
        diagnosis = random.choices(
            population=diagnoses,
            weights=[10, 10, 10, 10, 40, 10],  # 40% for 'Healthy'
            k=1
        )[0]

        chronic_condition = 'None' if diagnosis == 'Healthy' else random.choice(chronic_conditions)

        # Weighted reason for visit: 38% Checkup
        reason_for_visit = random.choices(
            population=visit_reasons,
            weights=[38, 12, 12, 12, 13, 13],  # 38% for 'Checkup'
            k=1
        )[0]

        record = {
            'Patient_ID': f'P{i:04d}',
            'First_Name': fake.first_name(),
            'Last_Name': fake.last_name(),
            'Date_of_Birth': dob.strftime('%Y-%m-%d'),
            'Gender': random.choice(genders),
            'Phone_Number': fake.phone_number(),
            'Email_Address': fake.email(),
            'Address': fake.street_address() + ", " + fake.city() + ", MD " + fake.zipcode(),  # Maryland only
            'Primary_Physician': random.choice(primary_physicians),
            'Visit_Date': visit_date.strftime('%Y-%m-%d'),
            'Reason_for_Visit': reason_for_visit,
            'Diagnosis': diagnosis,
            'Prescribed_Medication': random.choice(medications),
            'Allergies': random.choice(allergies),
            'Chronic_Conditions': chronic_condition,
            'Blood_Pressure': f"{random.randint(100, 140)}/{random.randint(60, 90)}",
            'Heart_Rate': random.randint(60, 100),
            'Lab_Results': f"WBC: {random.uniform(4.0, 11.0):.1f} x10^9/L",
            'Previous_Visits_Count': random.randint(0, 10),
            'Last_Visit_Date': last_visit.strftime('%Y-%m-%d'),
            'Follow_Up_Required': random.choice(['Yes', 'No']),
            'Next_Appointment_Date': next_appointment.strftime('%Y-%m-%d') if next_appointment else '',
            'Visit_Notes': fake.sentence(nb_words=10),
            'Insurance_Provider': random.choice(insurance_providers),
            'Policy_Number': fake.bothify(text='??#######') if random.random() < 0.9 else '',
            'Billing_Status': random.choice(billing_statuses),
            'Payment_Method': random.choice(payment_methods),
            'Record_Creation_Date': datetime.now().strftime('%Y-%m-%d'),
            'Last_Updated': datetime.now().strftime('%Y-%m-%d'),
            'Consent_Form_Signed': random.choice(['Yes', 'No']),
            'Risk_Level': random.choice(risk_levels),
            'Satisfaction_Score': random.randint(1, 10),
            'Treatment_Outcome': random.choice(treatment_outcomes)
        }
        records.append(record)
    return records

def save_to_csv(records, filename='patient_records_md_weighted.csv'):
    with open(filename, mode='w', newline='', encoding='utf-8') as file:
        writer = csv.DictWriter(file, fieldnames=records[0].keys())
        writer.writeheader()
        writer.writerows(records)
    return filename

# Example usage:
# data = generate_patient_data_with_weights()
# save_to_csv(data)
```
